```python
def read_event(fname):
    df = pandas.read_parquet(fname)
    event = df.iloc[0, 2]
    return df[['depth', 'cell_index']] \
        .set_index('cell_index') \
        .rename(columns={'depth': event})
    
event_df = read_event(fnames[0])
event_df
```

```python
event_dfs = [read_event(fname) for fname in fnames]
```

```python
len(fnames)
```

```python
all_events = pandas.concat(event_dfs, axis=1).fillna(0)
```
